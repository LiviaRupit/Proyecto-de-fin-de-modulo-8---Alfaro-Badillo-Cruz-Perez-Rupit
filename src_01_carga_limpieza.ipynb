```python
import pandas as pd

# 1. Cargar datos
sequia = pd.read_csv('../data/sequia_municipal_enero.csv')
v_amb = pd.read_csv('../data/vulnerabilidad_ambiental_municipal.csv')
v_eco = pd.read_csv('../data/vulnerabilidad_economica_municipal.csv')
v_soc = pd.read_csv('../data/vulnerabilidad_social_municipal.csv')

# 2. Homologar claves
v_amb = v_amb.rename(columns={'homologcve_concatenada': 'cve_concatenada'})

# 3. Seleccionar variables relevantes
sequia = sequia[['cve_concatenada', 'nombre_mun', 'entidad', 'msm_20260131']]
v_amb = v_amb[['cve_concatenada', 'probabilidad', 't_prob']]
v_eco = v_eco[['cve_concatenada', 'probabilidad', 't_prob']]
v_soc = v_soc[['cve_concatenada', 'probabilidad', 't_prob']]

# 4. Renombrar para claridad
v_amb.columns = ['cve_concatenada', 'prob_ambiental', 't_prob_ambiental']
v_eco.columns = ['cve_concatenada', 'prob_economica', 't_prob_economica']
v_soc.columns = ['cve_concatenada', 'prob_social', 't_prob_social']

# 5. Unir todo
df = sequia.merge(v_amb, on='cve_concatenada', how='left') \
           .merge(v_eco, on='cve_concatenada', how='left') \
           .merge(v_soc, on='cve_concatenada', how='left')

# 6. Guardar dataset final
df.to_csv('../data/dataset_final.csv', index=False)
```
