## Objetivo

*   Describir sequía y vulnerabilidades.
*   Generar gráficos y mapas.




```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('dataset_final.csv')
```

```python
# Histograma de sequía
sns.histplot(df['msm_20260131'], kde=True)
plt.title('Distribución de sequía meteorológica (msm_20260131)')
plt.savefig('hist_sequia.png')
```

```python
# Boxplots de vulnerabilidades
for col in ['t_prob_ambiental', 't_prob_economica', 't_prob_social']:
    sns.boxplot(x=df[col])
    plt.title(f'Distribución de {col}')
    plt.savefig(f'box_{col}.png')
```
